```python
import matplotlib.pyplot as plt

# 1. データを辞書で定義
fruits_data = {"apple":10, "banana":30, "orange":40, "kiwi":15}

# 2. 横軸の位置を決める（0, 1, 2, 3）
x = [1, 2, 3, 4]

# 3. グラフを描画
# fruits_data.values() で数字(10, 30...)、keys() で名前(apple...)を取得
plt.bar(x, fruits_data.values(), tick_label=list(fruits_data.keys()))

# 4. 表示（カッコを忘れずに！）
plt.show()
```
