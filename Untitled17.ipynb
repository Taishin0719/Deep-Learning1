```python
import random

def bubble_sort(lst):
  n = 0
  for j in range(len(lst) - 1):
    for i in range(len(lst) - 1 - j):
      n = n + 1
      if lst[i] > lst[i+1]:
        lst [i + 1], lst[i] = lst[i], lst[i+1]
  return n

def merge_sort_rec(data, l, r, work):
  if l+1 >= r:
    return 0
  m = l+(r-1)//2
  n1 = merge_sort_rec(data, l, m, work)
  n2 = merge_sort_rec(data, m, r, work)
  n = 0
  i1 = l
  i2 = m
  for i in range(l, r):
    from1 = False
    if i2 >= r:
      from1 = True
    elif i1 < m:
      n = n + 1
      if data[i1] <= data[i2]:
        from1 = True
    if from1:
      work[i] = data[i1]
      i1 = i1 + 1
  for i in range(l, r):
    data[i] = work[i]
  return n1+n2+n

def merge_sort(data):
  return merge_sort_rec(data, 0, len(data), [0]*len(data))
```
