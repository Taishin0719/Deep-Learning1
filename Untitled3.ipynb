```python
def greeting(en, fr, de):
  print(en + ", " + fr + ", " + de)

greeting("Hello", "Bonjour", "Guten Tag")
```

```python
def greeting(en, number, name):

  print(en*number+","+name)

greeting("Hello",3, "World")
```
