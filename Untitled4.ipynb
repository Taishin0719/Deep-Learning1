```python
def last_line(name):
    # 1. ファイルを読み込みモード("r")で開く
    with open(name, "r") as f:
        # 2. 行をすべて読み込んでリストにする
        lines = f.readlines()
        # 3. リストの最後[-1]を結果として返す
        return lines[-1]
```

```python
print(last_line("sample.txt")=="Excepteur sint sint occaecat cupidatat cupidatat non proident, sunt in culpa qui officia deserunt mollit anim id est laborum.＼n")
```

```python
def number_of_characters(name):
    with open(name, "r") as f:
        data = f.read()
        return len(data)
```

```python
print(number_of_characters("sample.txt") == 446)
```

```python
def file_upper(infile, outfile):
    # 1. 読み込み用のファイルを開く
    with open(infile, "r") as f_in:
        content = f_in.read()  # 中身を全部読み込む

    # 2. 文字列を大文字に変換する
    upper_content = content.upper()

    # 3. 書き込み用のファイルを開いて保存する
    with open(outfile, "w") as f_out:
        f_out.write(upper_content)
```

```python
with open("print-test.txt", "w") as f:
  print("hello", "world", file=f)
file_upper("print-test.txt", "print-test-upper.txt")
with open("print-test-upper.txt", "r") as f:
  print(f.read() == "HELLO WORLD＼n")
```

```python
# 文字コードを指定しないとWindowsならこちらでエラー
with open("utf-8.txt", "r", encoding="utf-8") as f:
  print(f.read())
```
