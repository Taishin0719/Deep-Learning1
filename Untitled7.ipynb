```python
class HelloFileIterator(HelloFile):
    def __iter__(self):
        return self

    def __next__(self):
        line = self.readline()
        if line == '':
            raise StopIteration
        return line

    # ↓ この2つを新しく追加！
    def __enter__(self):
        return self  # with ... as f の「f」に自分自身を渡す

    def __exit__(self, exc_type, exc_value, traceback):
        pass  # 特に後片付けが必要ない時は pass でOK
```

```python
with HelloFileIterator(3) as f:
  for line in f:
    print(line)
```
