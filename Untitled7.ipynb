```python
a = np.array([0,1,2,-3,-4,5,-6,-7])
print(a)
print(a[a < 0]) # 負の要素を取り出し
print(a[(a < 0) & (a % 2 == 0)]) # 負で偶数の要素を取り出し
a[a < 0] = 8 # 負の要素を8に上書き
print(a)
```

```python
def allsubstrings(s):
  return [s[i:j] for i in range(len(s)) for j in range(i+1,len(s)+1)]

allsubstrings("abc")
```
